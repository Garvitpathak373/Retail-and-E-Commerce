# E-Commerce Customer Loyalty Analytics - EDA

This notebook is a companion workspace for the project. The production EDA and model artifacts are generated by `train_model.py`.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

DATA_FILE = Path('../data/Ecommerce_Customer_Loyalty_Data.csv')
df = pd.read_csv(DATA_FILE)
df.head()
```

```python
df.info()
df.describe(include='all')
```

```python
numeric_df = df.select_dtypes(include='number')
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```
